```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Titanic dataset into the 'df' variable
df = sns.load_dataset('titanic')

# Bar Chart: Survival count by Passenger Class
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='class', hue='survived', palette='pastel')
plt.title('Survival Count by Passenger Class', fontsize=14)
plt.xlabel('Passenger Class', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.show()
```

```python
# Scatter Plot: Age vs Fare with Survival hue
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='age', y='fare', hue='survived', palette='deep', alpha=0.7)
plt.title('Age vs Fare (Survival Pattern)', fontsize=14)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Fare', fontsize=12)
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(10, 6))

# Sirf numerical columns ko select karke correlation nikalna
numeric_cols = df.select_dtypes(include=['float64', 'int64'])
corr_matrix = numeric_cols.corr()

# Heatmap generate karna custom colormap (cmap) ke sath
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap of Titanic Features', fontsize=14)
plt.show()
```
