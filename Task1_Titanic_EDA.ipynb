```python
import pandas as pd
import seaborn as sns

# Load the built-in Titanic dataset
df = sns.load_dataset('titanic')
```

```python
# View the first 5 rows
display(df.head())

# Check data types and non-null counts
df.info()

# Get statistical insights for numerical columns
display(df.describe())
```

```python
# Check for missing values
print(df.isnull().sum())

# Handle missing values (e.g., fill missing 'age' with the median, drop 'deck' because it has too many missing values)
df['age'] = df['age'].fillna(df['age'].median())
df.drop(columns=['deck'], inplace=True)

# Drop any duplicate rows
df = df.drop_duplicates()
```

```python
# Calculate average survival rate by gender
survival_by_gender = df.groupby('sex')['survived'].mean()
print("Survival by Gender:\n", survival_by_gender)

# Calculate average age and survival rate by passenger class
class_analysis = df.groupby('class')[['age', 'survived']].mean()
print("\nClass Analysis:\n", class_analysis)
```
