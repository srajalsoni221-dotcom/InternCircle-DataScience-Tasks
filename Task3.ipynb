```python
import pandas as pd
import numpy as np

# 1. Create a Mock E-commerce Dataset
np.random.seed(42)
dates = pd.date_range(start='2023-01-01', periods=365, freq='D')
categories = np.random.choice(['Electronics', 'Clothing', 'Home & Kitchen'], size=365)
revenue = np.random.uniform(50, 1000, size=365)
customer_type = np.random.choice(['New', 'Returning'], size=365)

df_sales = pd.DataFrame({
    'Order_Date': dates,
    'Category': categories,
    'Revenue': revenue,
    'Customer_Type': customer_type
})

# 2. Feature Engineering & Date Extraction
# Ensure Order_Date is datetime format, then extract month and day
df_sales['Order_Date'] = pd.to_datetime(df_sales['Order_Date'])
df_sales['Month'] = df_sales['Order_Date'].dt.month_name()
df_sales['Day_of_Week'] = df_sales['Order_Date'].dt.day_name()

print("Data after Feature Engineering:")
display(df_sales.head())
```

```python
# 3. Pivot Tables and Trend Calculations
# Calculate total revenue per category per month
pivot_revenue = pd.pivot_table(
    df_sales,
    values='Revenue',
    index='Month',
    columns='Category',
    aggfunc='sum'
)

# Sort months chronologically for better readability
months_order = ['January', 'February', 'March', 'April', 'May', 'June',
                'July', 'August', 'September', 'October', 'November', 'December']
pivot_revenue = pivot_revenue.reindex(months_order)

print("Revenue Trend (Pivot Table):")
display(pivot_revenue)
```
